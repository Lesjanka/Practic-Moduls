## 1. Классы и объекты
Класс — чертёж. Объект — вещь, построенная по этому чертежу. По одному чертежу можно сделать сколько угодно объектов.

```
class Sobaka:          <- чертёж (класс)
    ...

rex = Sobaka("Rex")    <- объект N1
bim = Sobaka("Bim")    <- объект N2, тот же чертёж
```

| Хочу | Пишу |
|---|---|
| Описать класс | `class Sobaka:` |
| Настройка при создании | `def __init__(self, ...):` |
| Поле объекта (данные) | `self.klichka = ...` |
| Создать объект | `rex = Sobaka("Rex")` |
| Метод объекта (действие) | `rex.golos()` |
| Проверить класс | `isinstance(rex, Sobaka)` |

`self` — это «сам объект»: через него метод видит свои поля.


````python
# Пример
class Sobaka:
    def __init__(self, klichka, vozrast):
        self.klichka = klichka
        self.vozrast = vozrast

    def golos(self):
        return self.klichka + ": gav!"

rex = Sobaka("Rex", 3)
bim = Sobaka("Bim", 1)
print(rex.golos())
print(bim.klichka, "|", bim.vozrast)
print("bim - это Sobaka?", isinstance(bim, Sobaka))
````

## 2. Инкапсуляция, атрибуты и свойства
Инкапсуляция — прячем внутренности объекта, наружу отдаём управляемый доступ.

```
объект
+------------------------------+
| imya          открытый       | <- доступ свободный
| _vozrast      с одним _      | <- договор: «внутренний, не лазь»
| __parol       с двумя __     | <- имя искажается: obj._Klass__parol
+------------------------------+
| property gradusy             | <- выглядит как поле,
|   @gradusy.setter            |    а внутри проверка
+------------------------------+
```

| Хочу | Пишу |
|---|---|
| Открытое поле | `self.imya = ...` |
| «Внутреннее» поле | `self._vozrast = ...` |
| Сильно скрытое поле | `self.__parol = ...` |
| Поле с проверкой | `@property` на чтение + `@имя.setter` на запись |


````python
# Пример
class Termometr:
    def __init__(self):
        self._gradusy = 20          # внутреннее поле

    @property
    def gradusy(self):              # чтение как обычное поле
        return self._gradusy

    @gradusy.setter
    def gradusy(self, znachenie):   # запись с проверкой
        if znachenie < -50 or znachenie > 60:
            raise ValueError("так не бывает")
        self._gradusy = znachenie

t = Termometr()
print(t.gradusy)      # 20
t.gradusy = 35        # проверка пропустила
print(t.gradusy)      # 35
try:
    t.gradusy = 999   # проверка не пропустила
except ValueError as oshibka:
    print("поймали:", oshibka)
````

## 3. Наследование
Класс-ребёнок получает поля и методы класса-родителя и добавляет свои.

```
          Zhivotnoe            <- родитель: klichka, spyat()
          /       \
     Sobaka       Kot          <- дети: свои методы
```

| Хочу | Пишу |
|---|---|
| Унаследовать | `class Sobaka(Zhivotnoe):` |
| Позвать конструктор родителя | `super().__init__(...)` |
| Родительский метод ребёнку | достаётся автоматически |
| Проверить родство | `issubclass(Sobaka, Zhivotnoe)` |


````python
# Пример
class Zhivotnoe:
    def __init__(self, klichka):
        self.klichka = klichka

    def spyat(self):
        return self.klichka + " spit"

class Sobaka(Zhivotnoe):
    def apport(self):
        return self.klichka + " nesit palochku"

s = Sobaka("Rex")          # своего __init__ нет - сработал родительский
print(s.spyat(), "|", s.apport())
print("Sobaka - потомок Zhivotnoe?", issubclass(Sobaka, Zhivotnoe))
````

## 4. Переопределение функционала базового класса
Ребёнок заменяет метод родителя: имя то же, код свой. Через `super().метод()` можно позвать родительскую версию и ДОПОЛНИТЬ её, а не выбрасывать.

| Хочу | Как |
|---|---|
| Заменить метод родителя | описать в ребёнке метод с ТЕМ ЖЕ именем |
| Позвать версию родителя | `super().metod()` |
| Дополнить, а не заменить | сначала `super().metod()`, потом свои строки |


````python
# Пример
class Figura:
    def ploshchad(self):
        return 0

    def opisanie(self):
        return "figura"

class Krug(Figura):
    def __init__(self, r):
        self.r = r

    def ploshchad(self):                    # замена
        return 3.14 * self.r ** 2

    def opisanie(self):                     # дополнение
        return super().opisanie() + ", krug"

k = Krug(2)
print(k.ploshchad(), "|", k.opisanie())
````

## 5. Атрибуты классов и статические методы
Атрибут класса ОДИН на все объекты (живёт в самом классе). Статический метод — функция внутри класса, которой не нужен объект.

```
class Student:
    vsego = 0              <- ОДИН на всех (атрибут класса)
    def __init__(self):
        self.imya = ...    <- у каждого объекта СВОЙ
        Student.vsego += 1
```

| Хочу | Пишу |
|---|---|
| Атрибут класса | имя = значение прямо в теле класса |
| Прочитать / поменять на всех | `Klass.imya` |
| Метод без self | `@staticmethod` |
| Метод с cls (видит класс) | `@classmethod` |


````python
# Пример
class Student:
    vsego_studentov = 0                 # атрибут класса

    def __init__(self, imya):
        self.imya = imya                # атрибут объекта
        Student.vsego_studentov += 1

    @staticmethod
    def sredniy_ball(bally):            # объекта не требует
        return sum(bally) / len(bally)

a = Student("Anna")
b = Student("Boris")
print("всего студентов:", Student.vsego_studentov)   # 2
print("средний:", Student.sredniy_ball([4, 5, 3]))   # 4.0
````

## 6. Класс object. Строковое представление объекта
Любой класс — потомок класса `object`, даже если родителей не писали. Оттуда приходят методы-«представители»:

| Хочу | Метод | Когда срабатывает |
|---|---|---|
| Строка «для людей» | `__str__` | `print(obj)`, `str(obj)` |
| Строка «для разработчика» | `__repr__` | объект в консоли, в списке, `repr(obj)` |
| Нет своих методов | достаётся вывод `object`: `<... object at 0x...>` |

Правило: `__str__` — красивый текст для пользователя, `__repr__` — точный текст для отладки (в идеале по нему можно воссоздать объект).


````python
# Пример
class Kniga:
    def __init__(self, nazvanie, avtor):
        self.nazvanie = nazvanie
        self.avtor = avtor

    def __str__(self):
        return self.nazvanie + " (" + self.avtor + ")"

    def __repr__(self):
        return "Kniga(" + repr(self.nazvanie) + ", " + repr(self.avtor) + ")"

k = Kniga("Kolobok", "narod")
print(k)        # сработал __str__
print([k])      # список показал __repr__
print("потомок object?", isinstance(k, object))
````

## 7. Перегрузка операторов
Волшебные методы учат собственный класс понимать `+`, `-`, `==`, `len()` и другие.

| Оператор / вызов | Метод |
|---|---|
| `a + b` | `__add__` |
| `a - b` | `__sub__` |
| `a * b` | `__mul__` |
| `a == b` | `__eq__` |
| `a < b` | `__lt__` |
| `len(a)` | `__len__` |
| `if a:` | `__bool__` |


````python
# Пример
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, drugoy):
        return Vector(self.x + drugoy.x, self.y + drugoy.y)

    def __eq__(self, drugoy):
        return (self.x, self.y) == (drugoy.x, drugoy.y)

    def __str__(self):
        return "(" + str(self.x) + "; " + str(self.y) + ")"

a = Vector(1, 2)
b = Vector(3, 4)
print(a + b)                      # (4; 6)
print(a + b == Vector(4, 6))      # True
````

## 8. Абстрактные классы и методы
Абстрактный класс — контракт: «объект меня создать нельзя, а дети обязаны реализовать эти методы».

| Хочу | Пишу |
|---|---|
| Абстрактный родитель | `class Figura(ABC):` |
| Метод-контракт | `@abstractmethod` |
| Ребёнок не реализовал контракт | TypeError при создании объекта ребёнка |


````python
# Пример
from abc import ABC, abstractmethod

class Figura(ABC):
    @abstractmethod
    def ploshchad(self):
        ...

class Kvadrat(Figura):
    def __init__(self, a):
        self.a = a

    def ploshchad(self):          # контракт выполнен
        return self.a ** 2

class Krug(Figura):
    def __init__(self, r):
        self.r = r

    def ploshchad(self):          # контракт выполнен
        return 3.14 * self.r ** 2

for f in [Kvadrat(2), Krug(1)]:
    print(f.ploshchad())
````

## Шпаргалка по части 3

| Тема | Главная мысль | Пример |
|---|---|---|
| Классы и объекты | чертёж и вещи по нему | `rex = Sobaka("Rex")` |
| Инкапсуляция | внутренности под замком, доступ через property | `@property`, `@x.setter` |
| Наследование | ребёнок берёт своё у родителя | `class Kot(Zhivotnoe):` |
| Переопределение | то же имя метода, свой код | `super().metod()` |
| Атрибуты класса | одно поле на все объекты | `Klass.schetchik += 1` |
| Строковое представление | `__str__` для людей, `__repr__` для отладки | `print(obj)` |
| Перегрузка операторов | свои `+`, `==`, `len()` | `__add__`, `__lt__` |
| Абстрактные классы | контракт: дети обязаны реализовать | `ABC`, `@abstractmethod` |
